Example:

If we have that X in our model has values (1: higher education, 0: no higher education)

And we get that the result of our model is

ln(P(Y=1)/P(Y=0)) = -3.91 + 0.24X

-3.91 is verything that X does not explain that affected the odds of defaulting/non-defaulting

If we transform the expression we get:

Odds (non-default | higher education) / Odds (non-default | non higher education) = e^0.24 = 1.27

That means, the odds of not defaulting having higher education are 1.27:1.00

Which means that if you have higher education yo have 27% more probability of non-defaulting than if you did not have higher education

### Let's checks the statistical significance of each dummy variable in the regression
### P-Values

In a regression, the p-value of a coefficient tells you how likely it is to observe a coefficient at least as large (in absolute value) as the one estimated, if the true coefficient were actually zero. Intuitively, it measures how much evidence the data provide that a predictor really has an impact on the target variable. A small p-value (e.g., below 0.05) suggests that the coefficient is unlikely to be different from zero just because of random sampling noise, so the variable is considered statistically significant. A large p-value suggests that the observed coefficient could easily have arisen by chance, meaning there is not enough evidence that the variable meaningfully contributes to the model.

For example, if a coefficient for annual_inc is 0.8 with a p-value of 0.01, it means there is strong evidence that income has a real relationship with the outcome. If the same coefficient had a p-value of 0.40, the estimated effect might simply be random noise, and we would be much less confident that income is truly predictive.

```python
from sklearn import linear_model
import scipy.stats as stat

class LogisticRegression_with_p_values:
    
    def __init__(self,*args,**kwargs):
        self.model = linear_model.LogisticRegression(*args,**kwargs)

    def fit(self,X,y):
        self.model.fit(X,y)
        denom = (2.0 * (1.0 + np.cosh(self.model.decision_function(X))))
        denom = np.tile(denom,(X.shape[1],1)).T
        F_ij = np.dot((X / denom).T,X)
        Cramer_Rao = np.linalg.inv(F_ij)
        sigma_estimates = np.sqrt(np.diagonal(Cramer_Rao))
        z_scores = self.model.coef_[0] / sigma_estimates
        p_values = [stat.norm.sf(abs(x)) * 2 for x in z_scores]
        self.coef_ = self.model.coef_
        self.intercept_ = self.model.intercept_
        self.p_values = p_values
```

```python
# Get the regression as an instance of the newly created logistic regression with p-values
reg = LogisticRegression_with_p_values()
```

```python
# Run the fit method
reg.fit(X=inputs_train, y=loan_data_output_train)
```

```python
# Overwrite the summary table

# Names of the columns
feature_names = inputs_train.columns.values
# Table with names and coefficients
summary_table = pd.DataFrame(columns=['Feature name'], data=feature_names)
summary_table['Coefficients'] = np.transpose(reg.coef_)
summary_table['P-Values'] = np.transpose(reg.p_values)

# ========== Let's add the intercept also ==============

# Move the index down of all rows in our table
summary_table.index = summary_table.index + 1
# Add in the first row the intercept, and there is no P-Value
summary_table.loc[0] = ['Intercept', reg.intercept_[0], np.nan]
# Sort by index
summary_table.sort_index(inplace=True)

# Let's add significance --- VECTORIZED
# We define a 1% significance
summary_table['Significance'] = np.where(
    (summary_table['P-Values'] <= 0.01) | (summary_table['P-Values'].isna()),
    'YES',
    'NO'
)

summary_table
```

If one or few dummy varibales per main variable (like grade for example) are statistically significant, it makes sense to retain all of them in the model, because otherwise we will loose information about variables. Only when all the subcategories of the variables are not significant can we comfortably decide to get rid of this variable as a whole.

```python
from math import ceil


def remove_weak_variables(summary_table):
    """
    Remove variables whose number of significant categories is too low.

    Rules:
    -------
    Categories are identified by the text before ':'.

    Examples:
        grade:A
        grade:B
        grade:C

    belong to the same variable: 'grade'.

    To keep a variable, it must have a minimum number of
    significant categories ('YES' in column 'Significance').

    Threshold:
        1-3 categories  -> at least 1 YES
        4-6 categories  -> at least 2 YES
        7-9 categories  -> at least 3 YES
        ...

    Formula:
        required_yes = ceil(number_of_categories / 3)

    Intercept is always preserved.

    Parameters
    ----------
    summary_table : pandas.DataFrame

    Returns
    -------
    pandas.DataFrame
        Filtered summary table.
    """

    # ------------------------------------------------------------------
    # Create copy to avoid modifying original dataframe
    # ------------------------------------------------------------------
    df = summary_table.copy()

    # ------------------------------------------------------------------
    # Identify intercept row(s)
    # ------------------------------------------------------------------
    intercept_mask = df['Feature name'] == 'Intercept'

    # ------------------------------------------------------------------
    # Extract variable name (text before ':')
    #
    # Example:
    #   grade:A  -> grade
    #   grade:B  -> grade
    # ------------------------------------------------------------------
    df['Variable'] = df['Feature name'].str.split(':').str[0]

    variables_to_keep = []

    # ------------------------------------------------------------------
    # Evaluate each variable independently
    # ------------------------------------------------------------------
    for variable, group in df.loc[~intercept_mask].groupby('Variable'):

        # Number of categories of the variable
        n_categories = len(group)

        # Number of significant categories
        n_yes = (group['Significance'] == 'YES').sum()

        # Minimum YES required
        required_yes = ceil(n_categories / 3)

        # Keep variable if threshold is met
        if n_yes >= required_yes:
            variables_to_keep.append(variable)

    # ------------------------------------------------------------------
    # Keep:
    #   - Intercept
    #   - Variables meeting the threshold
    # ------------------------------------------------------------------
    filtered_df = pd.concat([
        df.loc[intercept_mask],
        df.loc[
            (~intercept_mask) &
            (df['Variable'].isin(variables_to_keep))
        ]
    ])

    # ------------------------------------------------------------------
    # Remove helper column and reset index
    # ------------------------------------------------------------------
    filtered_df = filtered_df.drop(columns=['Variable'])
    filtered_df = filtered_df.reset_index(drop=True)

    return filtered_df
```

### `logistic_regression_model.predict(df_with_independent_variables)`

1. Multiplies the values of the independent variables by the model coefficients to calculate the **log-odds** of an observation being good.

2. Applies the exponential transformation to the log-odds to obtain the **odds** and the estimated **probability** of being good.

3. Applies a predefined **cut-off threshold** to the estimated probability and classifies the observation as **good** or **bad**.  --> Normally > 50% odds of being non defaulted will classify this as non-defaulted (1)

### Confussion matrix

A confusion matrix is a table that summarizes a classification model's predictions by comparing them with the actual outcomes. It shows the number of:

### ROC
As we can see, there is a trade off between getting less false positives and true positives

The ROC (Receiver Operating Characteristic) curve shows how a model's classification performance changes as the probability threshold used to assign classes is varied.

Intuitively: For every possible probability cutoff (e.g., 0.1, 0.2, 0.3, ..., 0.9), the ROC curve plots the True Positive Rate (Sensitivity) against the False Positive Rate (1 − Specificity). This allows us to assess the trade-off between correctly identifying positives and incorrectly classifying negatives as positives.

In credit risk, the ROC curve illustrates how the model's ability to distinguish between defaulters and non-defaulters changes as the PD threshold used for classification is adjusted.

### AUC (Area Under the Curve) is a metric that measures how well a classification model can distinguish between two classes. In credit risk modeling, it is usually the area under the ROC (Receiver Operating Characteristic) curve.

Intuitively: AUC represents the probability that the model assigns a higher risk score to a randomly chosen defaulted borrower than to a randomly chosen non-defaulted borrower.

AUC = 0.5 → No discriminatory power (equivalent to random guessing).
AUC = 1.0 → Perfect discrimination.
AUC < 0.5 → The model is systematically ranking observations in the wrong direction.

For example, an AUC of 0.80 means that in 80% of randomly selected pairs consisting of one defaulter and one non-defaulter, the model gives the higher probability of default to the defaulter.

In credit risk, AUC is widely used because it evaluates the model's ranking ability independently of the chosen probability cutoff.

### Gine Coefficient

The Gini coefficient is a measure of a model's ability to discriminate between two classes, commonly used in credit risk modeling. It is directly related to the AUC and is calculated as:

Gini=2×AUC−1

Intuitively: The Gini coefficient measures how much better the model is at ranking defaulters above non-defaulters compared to random guessing.

Gini = 0 → No discriminatory power (AUC = 0.5)
Gini = 1 → Perfect discrimination (AUC = 1.0)
Higher Gini → Better separation between defaulters and non-defaulters

For example, if a model has an AUC of 0.80:

Gini=2×0.80−1=0.60

So a Gini coefficient of 60% indicates strong discriminatory power and means the model is substantially better than random at ranking borrowers by credit risk.

### Kolmogorov-Smirnov coefficient

The Kolmogorov-Smirnov (KS) coefficient measures a model's ability to distinguish between two classes by calculating the maximum difference between the cumulative distributions of positives and negatives across all possible score or probability thresholds.

It is the maximum difference between cumulative distributions functions of 'good' and 'bad' borrowers.

Intuitively: The KS statistic indicates the point at which the separation between defaulters and non-defaulters is greatest. A higher KS means the model better differentiates high-risk from low-risk borrowers.

KS = 0% → No discriminatory power
Higher KS → Better discrimination
In credit risk, KS values above 30%-40% are often considered indicative of a good model.

For example, a KS of 45% means that at the optimal probability threshold, the cumulative proportion of defaulters and non-defaulters differs by 45 percentage points, showing strong separation between the two groups.

```python
# Plot KS
plt.plot(df_actual_predicted_probs['y_hat_test_proba'],df_actual_predicted_probs['Cumulative Perc Bad'], color='r')
# This is to see what would happen if we were guessing and classifying good and bad borrowers randomly
plt.plot(df_actual_predicted_probs['y_hat_test_proba'], df_actual_predicted_probs['Cumulative Perc Good'], color='b')
plt.xlabel('Estimated probability for being Good')
plt.ylabel('Cumulative %')
plt.title('Kolmogorov-Smirnov')

# Here the maximum distance becomes at 0.88-0.92 which means that this is the threshold at which the model achieves its highest accuracy
```

```python
KS = max(df_actual_predicted_probs['Cumulative Perc Bad'] - df_actual_predicted_probs['Cumulative Perc Good'])
KS
```

NOW HOW TO GET THE EXACT THRESHOLD POINT? A QUESTION TO TAKE WITH YOU

```python
def func_max_diff_AND_optimal_threshold_based_KS(
        df, # df_actual_predicted_probs
        cumulative_perc_bad_col, # 'Cumulative Perc Bad'
        cumulative_perc_good_col, # 'Cumulative Perc Good'
        probs_non_defaulting, # 'y_hat_test_proba'
        ):
    # df = df_actual_predicted_probs

    # Begin with the max set as 0
    max_difference_bad_good = 0

    # Initialize threshold
    threshold = None

    # Loop within the data frame
    for row in df.index:

        # Calculate the distance to get the maximum distance
        difference_bad_good = abs(
            df.loc[row, cumulative_perc_bad_col]
            - df.loc[row, cumulative_perc_good_col]
        )

        # Refresh the max
        if difference_bad_good > max_difference_bad_good:

            # Update the max value
            max_difference_bad_good = difference_bad_good

            # Get the value of probability of non-defaulting (being good)
            threshold = df.loc[row, probs_non_defaulting]

            # Continue to next row
            continue

    # If the difference has reached a maximum, then return
    # the difference and the threshold for that maximum
    # This assumes a unique maximum, which is not always the
    # case for the KS, but keeping the first occurrence is common
    # It can have multiple maximums at different points depending on the resulting function
    return max_difference_bad_good, threshold
```

```python
max_diff, threshold = func_max_diff_AND_optimal_threshold_based_KS(df=df_actual_predicted_probs,
                                                                   cumulative_perc_bad_col='Cumulative Perc Bad',
                                                                   cumulative_perc_good_col='Cumulative Perc Good',
                                                                   probs_non_defaulting='y_hat_test_proba')
print(f"The maximum difference {max_diff} shows a threshold of {threshold}")
```

```python
# We define a function called 'n_approved' which assigns a value of 1 if a predicted probability
# is greater than the parameter p, which is a threshold, and a value of 0, if it is not.
# Then it sums the column.
# Thus, if given any percentage values, the function will return
# the number of rows wih estimated probabilites greater than the threshold. 
def n_approved(p):
    return np.where(df_actual_predicted_probs['y_hat_test_proba'] >= p, 1, 0).sum()
```

```python
df_cutoffs['N Approved'] = df_cutoffs['thresholds'].apply(n_approved)
# Assuming that all credit applications above a given probability of being 'good' will be approved,
# when we apply the 'n_approved' function to a threshold, it will return the number of approved applications.
# Thus, here we calculate the number of approved appliations for al thresholds.
df_cutoffs['N Rejected'] = df_actual_predicted_probs['y_hat_test_proba'].shape[0] - df_cutoffs['N Approved']
# Then, we calculate the number of rejected applications for each threshold.
# It is the difference between the total number of applications and the approved applications for that threshold.
df_cutoffs['Approval Rate'] = df_cutoffs['N Approved'] / df_actual_predicted_probs['y_hat_test_proba'].shape[0]
# Approval rate equalts the ratio of the approved applications and all applications.
df_cutoffs['Rejection Rate'] = 1 - df_cutoffs['Approval Rate']
# Rejection rate equals one minus approval rate.
```
